```python
# ==================================
# Sistema Inteligente de Programadores
# ==================================

programadores = []

# FUNCION REGISTRAR
def registrar():

    nombre = input("Nombre: ")
    lenguaje = input("Lenguaje principal: ")
    repositorio = input("GitHub: ")
    proyectos = int(input("Proyectos realizados: "))
    experiencia = int(input("Años de experiencia: "))
    entorno = input("Entorno de desarrollo: ")
    seniority = input("Nivel (Junior/Mid/Senior): ")

    programadores.append({
        "Nombre": nombre,
        "Lenguaje": lenguaje,
        "Repositorio": repositorio,
        "Proyectos": proyectos,
        "Experiencia": experiencia,
        "Entorno": entorno,
        "Seniority": seniority
    })

    print("Programador agregado")


# FUNCION MOSTRAR
def mostrar():

    if len(programadores) == 0:
        print("No hay registros")

    else:
        print("\n--- LISTA DE PROGRAMADORES ---")

        for p in programadores:

            print("\nNombre:", p["Nombre"])
            print("Lenguaje:", p["Lenguaje"])
            print("GitHub:", p["Repositorio"])
            print("Proyectos:", p["Proyectos"])
            print("Experiencia:", p["Experiencia"])
            print("Entorno:", p["Entorno"])
            print("Nivel:", p["Seniority"])

            # IA SIMPLE
            if p["Experiencia"] >= 5:
                print("IA: Programador con experiencia")

            elif p["Experiencia"] >= 2:
                print("IA: Nivel intermedio")

            else:
                print("IA: Programador principiante")


# MENU
while True:

    print("\n1. Registrar")
    print("2. Mostrar")
    print("3. Salir")

    opcion = input("Opcion: ")

    if opcion == "1":
        registrar()

    elif opcion == "2":
        mostrar()

    elif opcion == "3":
        print("Programa finalizado")
        break

    else:
        print("Programa realizado por Marco Moreno NL=16")
```
